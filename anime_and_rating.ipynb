```python
# prompt: fine tune the above model

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, Flatten, Dense, Concatenate, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from sklearn.metrics import accuracy_score, mean_squared_error
from sklearn.preprocessing import StandardScaler

# Load data (assuming files are in your Google Drive)
rating = pd.read_csv('/content/drive/MyDrive/anime and rating data/rating.csv')
anime = pd.read_csv('/content/drive/MyDrive/anime and rating data/anime.csv')

# Merge datasets
df = pd.merge(rating, anime.drop('rating', axis=1), on='anime_id')

# Data cleaning
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
df.drop_duplicates(subset=['name'], keep='first', inplace=True)
df['name'] = df['name'].str.strip()

# Feature engineering
df['episodes'] = pd.to_numeric(df['episodes'], errors='coerce').fillna(0) # Fill NaN with 0
df['genre'] = df['genre'].astype(str)
df['type'] = df['type'].astype(str)
df['name'] = df['name'].astype(str)

# Rating conversion (improved)
df['rating'] = df['rating'].clip(lower=1, upper=10)
df['rating'] = df['rating'].astype(int)


# Data scaling
scaler = StandardScaler()
df['episodes'] = scaler.fit_transform(df[['episodes']])


# Prepare data for the model
user_ids = df['user_id'].unique()
anime_ids = df['anime_id'].unique()
num_users = len(user_ids)
num_animes = len(anime_ids)

user_mapping = {user_id: i for i, user_id in enumerate(user_ids)}
anime_mapping = {anime_id: i for i, anime_id in enumerate(anime_ids)}

df['user_index'] = df['user_id'].map(user_mapping)
df['anime_index'] = df['anime_id'].map(anime_mapping)

# Split data
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Model building
embedding_size = 128

user_input = Input(shape=(1,), name='user_input')
anime_input = Input(shape=(1,), name='anime_input')

user_embedding = Embedding(num_users, embedding_size, embeddings_regularizer=l2(1e-6))(user_input)
anime_embedding = Embedding(num_animes, embedding_size, embeddings_regularizer=l2(1e-6))(anime_input)

user_flat = Flatten()(user_embedding)
anime_flat = Flatten()(anime_embedding)
merged = Concatenate()([user_flat, anime_flat])

dense_1 = Dense(512, activation='relu')(merged)
dense_1 = Dropout(0.3)(dense_1)
dense_2 = Dense(256, activation='relu')(dense_1)
dense_2 = Dropout(0.3)(dense_2)
output = Dense(1, activation='linear')(dense_2) # Linear activation for regression

model = Model(inputs=[user_input, anime_input], outputs=output)
model.compile(optimizer=Adam(learning_rate=0.001), loss='mse', metrics=['mae']) # Added MAE metric
model.summary()

# Model training
history = model.fit([train_df['user_index'], train_df['anime_index']], train_df['rating'],
                    epochs=50, batch_size=256, validation_split=0.1)


# Model evaluation
predictions = model.predict([test_df['user_index'], test_df['anime_index']])
mse = mean_squared_error(test_df['rating'], predictions)
print(f"Mean Squared Error: {mse}")

# You can also plot the training history for further analysis
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')
plt.show()
```

```python
def recommend_anime(user_id, num_recommendations=10):
    # Get the index of the user from the user_mapping
    user_index = user_mapping.get(user_id)

    # Get a list of anime IDs the user has not watched
    watched_anime = df[df['user_id'] == user_id]['anime_id'].tolist()
    unwatched_anime = list(set(anime_ids) - set(watched_anime))

    if user_index is not None:
        # Prepare inputs for prediction
        user_inputs = np.array([user_index] * len(unwatched_anime))
        anime_inputs = np.array([anime_mapping[anime_id] for anime_id in unwatched_anime])

        # Reshape the inputs to have a defined shape
        user_inputs = user_inputs.reshape(-1, 1)  # Reshape to (number_of_samples, 1)
        anime_inputs = anime_inputs.reshape(-1, 1)  # Reshape to (number_of_samples, 1)

        # Predict ratings
        predicted_ratings = model.predict([user_inputs, anime_inputs]).flatten()

        # Create a DataFrame for recommendations
        recommendations = pd.DataFrame({
            'anime_id': unwatched_anime,
            'predicted_rating': predicted_ratings
        })

        # Sort by predicted rating in descending order
        recommendations = recommendations.sort_values(by='predicted_rating', ascending=False).head(num_recommendations)

        # Merge with anime metadata for additional details
        recommendations = recommendations.merge(anime[['anime_id', 'name', 'genre', 'type']], on='anime_id')

        return recommendations
    else:
        print(f"User ID {user_id} not found in the user mapping.")
        return None

# Example usage
user_id = 1  # Replace with a valid user ID
num_recommendations = 10
recommended_anime = recommend_anime(user_id, num_recommendations)

if recommended_anime is not None:
    print("Top Recommendations:")
    print(recommended_anime)
```

```python
from implicit.als import AlternatingLeastSquares
import scipy.sparse as sparse
import numpy as np
from sklearn.metrics import mean_squared_error

# Create a user-item interaction matrix
user_item_matrix = df.pivot_table(index='user_id', columns='anime_id', values='rating').fillna(0)

# Initialize the ALS model
model1 = AlternatingLeastSquares(factors=50, regularization=0.1, iterations=15)

# Convert the user-item matrix to a sparse CSR matrix
sparse_user_item = sparse.csr_matrix(user_item_matrix.T)

# Fit the ALS model (use model1 instead of model)
model1.fit(sparse_user_item)

predictions = []
for _, row in test_df.iterrows():
    user_id = row['user_id']
    anime_id = row['anime_id']
    if user_id in user_item_matrix.index and anime_id in user_item_matrix.columns:
        user_idx = user_item_matrix.index.get_loc(user_id)
        anime_idx = user_item_matrix.columns.get_loc(anime_id)
        # Use model1 for recommendations
        recommendations = model1.recommend(user_idx, sparse_user_item[anime_idx], N=1)
        prediction = recommendations[0][0]
        predictions.append(prediction)
    else:
        predictions.append(np.mean(df['rating']))

threshold = 5
predicted_labels = (np.array(predictions) > threshold).astype(int)
actual_labels = (test_df['rating'] > threshold).astype(int)
accuracy = accuracy_score(actual_labels, predicted_labels)
print(f'Test Accuracy: {accuracy}')
```

```python
def recommend_anime(user_id, top_n=10):
  # Get the user's preferred genres and average rating.
  user_ratings = df[df['user_id'] == user_id]
  user_avg_rating = user_ratings['rating'].mean()
  user_genres = user_ratings['genre'].str.split(',').explode().value_counts()

  # Filter anime based on the user's preferences.
  recommended_anime = df[~df['anime_id'].isin(user_ratings['anime_id'])]  # Exclude anime already rated by user
  recommended_anime = recommended_anime[recommended_anime['rating'] >= user_avg_rating]
  recommended_anime = recommended_anime[recommended_anime['genre'].str.contains('|'.join(user_genres.index[:3]))]

  # Sort the recommendations by rating and return the top N.
  recommended_anime = recommended_anime.groupby('name')['rating'].mean().sort_values(ascending=False).head(top_n).index.tolist()

  return recommended_anime

# Example usage
user_id_to_recommend = 1
recommendations = recommend_anime(user_id_to_recommend)
print(f"{recommendations}")
```

```python
def recommend_anime_with_details(user_id, top_n=10):

  recommendations = recommend_anime(user_id, top_n)

  recommended_anime_data = []
  for anime_name in recommendations:
    anime_row = df[df['name'] == anime_name].iloc[0]
    recommended_anime_data.append({'Name': anime_name,'Genre': anime_row['genre'],'Rating': anime_row['rating']})

  return pd.DataFrame(recommended_anime_data)

# Example usage
user_id_to_recommend = 1
recommendations_with_details = recommend_anime_with_details(user_id_to_recommend)
recommendations_with_details
```
